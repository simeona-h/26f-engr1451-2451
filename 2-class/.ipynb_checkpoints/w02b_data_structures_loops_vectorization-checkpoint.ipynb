## 1. Choose the right data structure

| Structure | Example | Ordered? | Mutable? | Typical use |
|---|---|---:|---:|---|
| List | `[1, 2, 3]` | Yes | Yes | A sequence that may change |
| Tuple | `(1, 2, 3)` | Yes | No | A fixed group of related values |
| Dictionary | `{"id": "A1"}` | Yes* | Yes | Labeled key–value data |
| Set | `{1, 2, 3}` | No positional indexing | Yes | Unique values and membership tests |

\*Dictionaries preserve insertion order in modern Python, but values are normally accessed by **key**, not by position.

Two syntax details:

- a one-item tuple needs a comma: `(5,)`;
- an empty set is `set()`, because `{}` creates an empty dictionary.

````python
measurements = [12.4, 10.8, 13.1]             # list
coordinates = (40.44, -79.96)                 # tuple
sample = {"id": "A1", "passed": True}       # dictionary
materials = {"Al", "Cu", "Al"}              # set removes duplicates

print(measurements)
print(coordinates)
print(sample)
print(materials)
````

## 2. Accessing and changing data

### Lists

List indexing starts at `0`. Negative indices count backward from the end, and a slice selects a range of values. Lists are **mutable**, so their contents can change.

Examples: measurements, names, tasks, files

````python
measurements = [12.4, 10.8, 13.1, 9.9, 11.7]

print("first:", measurements[0])
print("last:", measurements[-1])
print("middle slice:", measurements[1:4])

measurements[1] = 10.9
measurements.append(12.0)

average = sum(measurements) / len(measurements)
print("updated:", measurements)
print("average:", average)
````

### Tuples, dictionaries, and sets

- A **tuple** supports indexing and slicing, but its values cannot be replaced. Examples: coordinates, RGB colors, dimensions
- A **dictionary** stores values under descriptive keys. Examples: student info, settings, material properties
- A **set** stores unique values and is especially useful with `in` and `not in`. Examples: unique categories, allowed options, overlap

````python
point = (3, 5)
print("x coordinate:", point[0])

sample = {"id": "A1", "temperature_c": 22.5}
print("sample ID:", sample["id"])
sample["passed"] = True                        # add a key-value pair
sample["temperature_c"] = 23.0                 # update a value
print("updated dictionary:", sample)

materials = {"Al", "Cu"}
materials.add("Ag")
print("Cu included?", "Cu" in materials)
print("unique materials:", materials)
````

## 3. Decisions with `if`

A condition evaluates to either `True` or `False`. Indentation identifies the code belonging to each branch.

```python
if condition_1:
    # runs when condition_1 is True
elif condition_2:
    # runs when condition_2 is True
else:
    # runs otherwise
```

````python
temperature_c = 22

if temperature_c < 10:
    description = "cold"
elif temperature_c <= 25:
    description = "comfortable"
else:
    description = "hot"

print(description)
````

## 4. Repetition with `for` loops

A `for` loop runs once for each item in a sequence. Use a descriptive temporary name and indent the loop body by **4 spaces**.

````python
measurements = [12.4, 10.8, 13.1, 9.9, 11.7]
squared_measurements = []

for value in measurements:
    squared_measurements.append(value ** 2)

squared_measurements
````

Two common loop helpers are:

- `range(start, stop)`, which generates integers up to—but not including—`stop`;
- `enumerate(sequence)`, which provides both the index and the value.

````python
for trial in range(1, 4):
    print("trial", trial)

print("---")

for index, value in enumerate(measurements):
    print(index, value)
````

Loops and conditions are often combined to select or count values.

````python
threshold = 12.0
above_threshold = []

for index, value in enumerate(measurements):
    if value > threshold:
        above_threshold.append((index, value))

above_threshold
````

## 5. Vectorization: operating on an entire array

A Python `for` loop handles one item at a time. **Vectorized** code expresses an operation for an entire NumPy array at once.

Vectorization is often preferred for numerical data because it is:

- shorter and closer to the mathematical expression;
- usually faster for large arrays; and
- less likely to contain indexing or bookkeeping errors.

First import NumPy and convert the list to an array.

````python
import numpy as np

measurements_array = np.array(measurements)
measurements_array
````

### The same calculation two ways

The loop and vectorized expressions below both square every measurement.

````python
# Explicit Python loop
squared_with_loop = []
for value in measurements:
    squared_with_loop.append(value ** 2)

# Vectorized NumPy operation
squared_vectorized = measurements_array ** 2

print("loop:", squared_with_loop)
print("vectorized:", squared_vectorized)

print(np.allclose(squared_with_loop, squared_vectorized))
````

### Replacing a loop and `if` with a Boolean mask

A comparison involving an array produces one Boolean value per element. This Boolean array is called a **mask** and can be used to select matching values.

`np.flatnonzero()` returns the indices of elements that are nonzero.

````python
threshold = 12.0
mask = measurements_array > threshold

values_above = measurements_array[mask]
indices_above = np.flatnonzero(mask)

print("mask:", mask)
print("indices:", indices_above)
print("values:", values_above)
````

### Does vectorization speed up the calculation?

The following comparison performs the same numerical calculation on many values. The data structures are created **before** timing so that the comparison focuses on the calculation itself.

The exact times will vary by computer and from run to run. For small collections, the difference may not matter, and converting data to a NumPy array also has a cost.

````python
from time import perf_counter

n = 500_000
large_list = list(range(n))
large_array = np.arange(n, dtype=float)

start = perf_counter()
loop_result = []
for value in large_list:
    loop_result.append(value ** 2 + 3 * value)
loop_time = perf_counter() - start

start = perf_counter()
vectorized_result = large_array ** 2 + 3 * large_array
vectorized_time = perf_counter() - start

speedup = loop_time / vectorized_time

print(f"Loop time:       {loop_time:.4f} seconds")
print(f"Vectorized time: {vectorized_time:.4f} seconds")
print(f"Speedup:         {speedup:.1f}×")
print("Same values?", np.allclose(loop_result, vectorized_result))
````

## 6. Compact code reference

| Goal | Python pattern |
|---|---|
| Create a list | `values = [3, 5, 7]` |
| Create a tuple | `point = (3, 5)` |
| Create a dictionary | `sample = {"id": "A1", "value": 12.4}` |
| Create a set | `groups = {"control", "treatment"}` |
| First / last item | `values[0]`, `values[-1]` |
| Slice items | `values[start:stop]` |
| Replace / append a list item | `values[i] = x`, `values.append(x)` |
| Length / summary | `len(values)`, `min(values)`, `max(values)`, `sum(values)` |
| Dictionary lookup / update | `sample["id"]`, `sample["passed"] = True` |
| Membership test | `item in container` |
| Conditional branches | `if ...:`, `elif ...:`, `else:` |
| Loop over values | `for value in values:` |
| Loop over integers | `for i in range(start, stop):` |
| Loop over index and value | `for i, value in enumerate(values):` |
| Convert to a NumPy array | `array = np.array(values)` |
| Vectorized arithmetic | `result = array * 2 + 1` |
| Create a Boolean mask | `mask = array > limit` |
| Select matching values | `array[mask]` |
| Find matching indices | `np.flatnonzero(mask)` |
| Compare numerical results | `np.allclose(a, b)` |
| Compare exact arrays | `np.array_equal(a, b)` |

## 8. Key takeaways

- Lists are the default choice for an ordered sequence that may change.
- Tuples hold fixed ordered values, dictionaries hold labeled values, and sets hold unique values.
- `if` statements choose what code runs; `for` loops repeat code across a sequence.
- NumPy vectorization applies numerical operations to entire arrays without an explicit Python loop.
- Boolean masks provide a vectorized alternative to many loop-and-`if` filtering tasks.
- Vectorization is often substantially faster for large numerical arrays, but clarity and problem size still matter.
