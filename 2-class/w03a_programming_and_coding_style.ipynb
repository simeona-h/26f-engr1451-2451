## 1. Plan before you code

> “The sooner you start to code, the longer the program will take.” — Roy Carlson

Before writing Python, describe the steps in plain language. This is **pseudocode**.

### Example problem
Calculate the mean of a set of measurements and report whether it exceeds a threshold.

### Pseudocode
1. Store the measurements.
2. Calculate their mean.
3. Compare the mean with the threshold.
4. Print the result.

````python
measurements = [12.1, 11.8, 12.5, 12.0]
threshold = 12.0

mean_measurement = sum(measurements) / len(measurements)
is_above_threshold = mean_measurement > threshold

print(f"Mean: {mean_measurement:.2f}")
print(f"Above threshold: {is_above_threshold}")
````

## 2. Why readable code matters

Readable code is easier to:

- grade and review,
- debug,
- modify later, and
- share with teammates.

### Spaghetti code
**Spaghetti code** is tangled, repetitive, or difficult to follow.

### Code smell
A **code smell** is a feature of code that may indicate a deeper problem, such as:

- unclear names,
- duplicated code,
- very long functions,
- deeply nested conditionals,
- unexplained constants,
- inconsistent formatting.

## 3. Comments: explain *why*, not the obvious

Useful comments can describe:

- the purpose of a calculation,
- assumptions about data,
- units,
- unusual design decisions,
- error handling,
- a non-obvious algorithm,
- something that may need to change later.

Avoid comments that merely repeat the code.

````python
# Not useful: the code already says this.
x = 3 + 5  # Add 3 and 5 and assign the result to x

# Better: the comment explains information that is not obvious from the code.
temperature_c = 25  # Laboratory test temperature
````

### Comments are not deodorant

A comment should not be used to rescue a confusing variable name.

**Less clear**
```python
l = 30  # latitude in degrees
```

**Better**
```python
latitude_deg = 30
```

Often, improving the code removes the need for a comment.

## 4. Use consistent indentation

Python uses indentation to define blocks of code. The standard convention is **4 spaces per indentation level**.

Consistent indentation makes nested `if` statements, loops, and functions much easier to follow.

````python
x = 8
y = 2

if y == 0:
    result = None
else:
    result = x ** y

result
````

## 5. Write Boolean logic directly

Avoid comparisons such as:

```python
if flag == True:
```

Use:

```python
if flag:
```

Also avoid an entire `if/else` block when the comparison itself already produces `True` or `False`.

````python
matches = 3

# Longer than necessary
if matches > 0:
    found = True
else:
    found = False

print(found)

# Better
found = matches > 0
print(found)
````

## 6. Naming in Python

For this course, follow common Python conventions:

- variables: `snake_case`
- functions: `snake_case`
- constants: `UPPER_CASE`
- classes: `CamelCase`

Use names that communicate meaning.

### Prefer
```python
num_errors
num_completed_connections
compute_arrival_time()
is_overpriced
has_license
```

### Avoid
```python
nerr
ncc
comp_arr()
x1
temp2
```

A short name such as `i` is reasonable for a very small loop, but descriptive iterator names are often clearer.

````python
file_names = ["data1.csv", "data2.csv", "data3.csv"]

for file_name in file_names:
    print(file_name)
````

### Naming Boolean variables

Boolean names should read naturally as `True` or `False`.

Prefer:

- `is_found`
- `has_data`
- `can_execute`
- `should_sort`

Avoid double negatives such as `is_not_missing` when a simpler positive form is possible.

## 7. Functions

Functions help organize repeated or logically related work.

A good function should:

- have one clear purpose,
- use a descriptive name,
- accept inputs rather than depend unnecessarily on global variables,
- return a useful result when appropriate.

In Python, use an explicit `return` when a function produces an output.

````python
def add_values(x, y):
    return x + y


add_values(3, 5)
````

## 8. Make units explicit

Units are part of the meaning of a variable.

**Mars Climate Orbiter failure (1999):** Lockheed Martin and NASA/JPL teams used different units for thruster impulse (**lbf·s** vs. **N·s**). The resulting navigation error caused the orbiter to enter the Martian atmosphere much lower than intended, and the spacecraft was lost.

**Cost:** approximately $125 million.

Prefer:

```python
incident_angle_deg = 90
length_cm = 35
time_seconds = 10
```

rather than:

```python
angle = 90
length = 35
time = 10
```

For simple course work, putting the unit in the variable name is often sufficient.

````python
distance_m = 250
time_s = 20

speed_m_per_s = distance_m / time_s
speed_m_per_s
````

### Optional: unit-aware quantities

For larger engineering projects, a units library such as `pint` can help prevent mistakes. It is not required for this course.

from pint import UnitRegistry

x + y

## 9. Avoid magic numbers

A **magic number** is a number placed directly in a calculation without explaining what it represents.

### Harder to understand
```python
students = 4 * 15
```

### Better
```python
students_per_group = 4
num_groups = 15
students = students_per_group * num_groups
```

Named values document the designer's intent and make later changes easier.

````python
students_per_group = 4
num_groups = 15

num_students = students_per_group * num_groups
num_students
````

## 10. Eliminate code duplication

Repeated blocks of code are harder to maintain because a change must be made everywhere.

If you find yourself copying and pasting the same logic, consider creating a function.

**Once and only once:** each important piece of logic should ideally have one authoritative implementation.

````python
values_a = [2, 4, 6, 8]
values_b = [3, 6, 9, 12]

def summarize(values):
    return {
        "mean": sum(values) / len(values),
        "min": min(values),
        "max": max(values),
    }

print(summarize(values_a))
print(summarize(values_b))
````

## 11. Prefer versatile functions

A useful function should work for more than one hard-coded case.

### Less versatile
```python
def mean_group_a():
    return sum(group_a) / len(group_a)
```

### More versatile
```python
def mean_value(values):
    return sum(values) / len(values)
```

The second function can be tested and reused with many datasets.

## 13. Refactoring example: repeated analysis

Suppose we want the 3 most common words used by several artists.

Repeated code is a signal that a function may help.

````python
from collections import Counter
import re

lyrics_by_artist = {
    "artist_a": "data science is fun and data is useful",
    "artist_b": "python is useful and python is readable",
}

def top_words(text, num_words=3):
    words = re.findall(r"[a-z']+", text.lower())
    return Counter(words).most_common(num_words)

for artist, lyrics in lyrics_by_artist.items():
    print(artist, top_words(lyrics))
````

## 14. Refactoring example: reduce nested conditionals

Deeply nested `if` statements are difficult to read and are a common code smell.

Suppose:

- baseline or HF exposure gives dose `0`,
- step `0` also gives dose `0`,
- another exposure accumulates dose over steps.

Rather than writing a different formula for every step, isolate the repeated calculation.

````python
def calculate_dose(exposure, step, exposure_values, scale_factor):
    if exposure in {"baseline", "HF"} or step == 0:
        return 0.0

    if exposure == "mASTMG154":
        return sum(exposure_values[:step]) * scale_factor

    return None


exposure_values = [10, 12, 11, 13]
scale_factor = 0.21835 * 8 / 12

calculate_dose(
    exposure="mASTMG154",
    step=3,
    exposure_values=exposure_values,
    scale_factor=scale_factor,
)
````

### Why this version is easier to maintain

- repeated arithmetic appears once,
- the exposure logic is easier to see,
- the function can be tested independently,
- the calculation works for additional steps without another `if` statement.

## 16. Compact Python coding-style reference

| Topic | Prefer | Avoid |
|---|---|---|
| Variable names | `mean_temperature` | `mt`, `x2` |
| Function names | `compute_mean()` | `ComputeMean()`, `cmpmn()` |
| Boolean names | `is_valid`, `has_data` | `flag1`, `is_not_invalid` |
| Constants | `MAX_ITERATIONS = 100` | unexplained `100` |
| Indentation | 4 spaces | inconsistent indentation |
| Boolean tests | `if is_valid:` | `if is_valid == True:` |
| Comments | explain **why** | restate the code |
| Units | `distance_m` | `distance` |
| Repeated logic | function / loop | copy-and-paste |
| DataFrame work | column names / vectorized operations | unnecessary row-by-row loops |
| Functions | one clear purpose | long function doing many unrelated tasks |
| Notebook cells | one logical task | huge all-in-one cells |
